```python
import numpy as np
import qutip as qt
#I suggest that create a new environment for qutip just in case.
#But I think it shoud be alright if you don't.
```

```python
def random_state(nbasis):
    real_parts = np.array([])
    im_parts = np.array([])
    for amplitude in range(nbasis):
        real_parts = np.append(real_parts, (np.random.random()*2)-1)
        im_parts = np.append(im_parts, (np.random.random()*2)-1)
    # Combine into list of complex numbers:
    amps = real_parts + 1j*im_parts
    # Normalise
    magnitude_squared = 0
    for a in amps:
        magnitude_squared += abs(a)**2
    amps /= np.sqrt(magnitude_squared)
    return amps
```

```python
#You can set dimension by change the value of d.
#The value of d should at least 2.
d = 3
```

```python
#Generate a random state
psi = qt.Qobj(random_state(d))
psi = psi*psi.dag()
psi
```

```python
#Generate d dimention entangled state
PureState = []
for i in range(d):
    PureState.append(qt.basis(d, i))

PureEntangled = qt.tensor(PureState[0], PureState[0])
for i in range(1, d):
    PureEntangled = PureEntangled + qt.tensor(PureState[i], PureState[i])
PureEntangled = np.sqrt(1/d)*PureEntangled
DenEntangled = PureEntangled*PureEntangled.dag()
DenEntangled
```

```python
#Teleportation
(qt.tensor(psi, DenEntangled))*qt.tensor(DenEntangled, qt.qeye(d))
```

```python
#Do partial trace
(qt.tensor(psi, DenEntangled)*qt.tensor(DenEntangled, qt.qeye(d))).ptrace(2)
```
